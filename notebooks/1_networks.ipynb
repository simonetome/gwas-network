```python
import pandas as pd
import networkx as nx
import pickle
```

Objective: read all network and save them in a networkx format

```python
# raw data path
data_path="/home/ubuntu/Data/"
```

```python
# string filtered
string = pd.read_parquet("../data/quality_control/string_filtered.parquet")
string = string[~(string['alias_sym1'].isna())]
string = string[~(string['alias_sym2'].isna())].reset_index(drop=True)
string.head()
```

```python
string_nx = nx.from_pandas_edgelist(
    df=string,
    source='alias_sym1',
    target='alias_sym2'
)
```

```python
with open("../data/network/string.pkl","wb") as out_f:
    pickle.dump(string_nx, out_f)
```

```python
# bioplex 
bioplex = pd.read_csv(data_path+"networks/BioPlex_293T_Network_10K_Dec_2019.tsv", sep="\t")
bioplex.head()
```

```python
bioplex_nx = nx.from_pandas_edgelist(
    df=bioplex,
    source='SymbolA',
    target='SymbolB'
)
```

```python
with open("../data/network/bioplex.pkl","wb") as out_f:
    pickle.dump(bioplex_nx, out_f)
```

```python
biogrid = pd.read_csv(data_path+"networks/BIOGRID-MV-Physical-5.0.258.tab3.txt", sep="\t")
biogrid = biogrid[biogrid['Organism Name Interactor A'] == 'Homo sapiens']
biogrid = biogrid[biogrid['Organism Name Interactor B'] == 'Homo sapiens']
biogrid.head(1)
```

```python
biogrid_nx = nx.from_pandas_edgelist(
    df=biogrid,
    source='Official Symbol Interactor A',
    target='Official Symbol Interactor B'
)
```

```python
with open("../data/network/biogrid.pkl","wb") as out_f:
    pickle.dump(biogrid_nx, out_f)
```

```python
humannet = pd.read_csv(data_path+"networks/HumanNet-FN.symbol.tsv", sep="\t", header=None)
humannet.columns = ['GeneA','GeneB','Score']
humannet.head(1)
```

```python
humannet_nx = nx.from_pandas_edgelist(
    df=humannet,
    source='GeneA',
    target='GeneB'
)
```

```python
with open("../data/network/humannet.pkl","wb") as out_f:
    pickle.dump(humannet_nx, out_f)
```

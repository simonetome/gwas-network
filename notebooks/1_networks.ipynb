```python
import pandas as pd
import networkx as nx
import pickle
```

Objective: read all network and save them in a networkx format

```python
# raw data path
data_path="/home/ubuntu/Data/"
```

```python
# string filtered
string = pd.read_parquet("../data/quality_control/string_physical_filtered.parquet")
string = string[~(string['alias_sym1'].isna())]
string = string[~(string['alias_sym2'].isna())].reset_index(drop=True)
string.head()
```

```python
string_nx = nx.from_pandas_edgelist(
    df=string,
    source='alias_sym1',
    target='alias_sym2'
)
```

```python
with open("../data/network/string_physical.pkl","wb") as out_f:
    pickle.dump(string_nx, out_f)
```

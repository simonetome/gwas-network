```python
import pandas as pd
import pickle
```

```python
related_open_targets = pd.read_parquet("../data/quality_control/open_target_annotated_subset.parquet")
related_open_targets.head(1)
```

```python
top_genes = pd.read_csv("../output/top_1perc_genes_RRA.csv", index_col=0)
top_genes.head(2)
```

```python
related_open_targets = (
    related_open_targets
    .merge(
        top_genes,
        how='inner',
        left_on='Approved symbol',
        right_index=True,
    )
)
```

```python
related_open_targets = related_open_targets[related_open_targets['score'] >= 0.1]
```

```python
related_open_targets.to_csv("../output/ot_related_topGenes.csv")
```

```python
import pandas as pd
from gprofiler import GProfiler
import re
```

```python
top_genes = pd.read_csv("../output/top_1perc_genes_RRA.csv", index_col=0)
top_genes.head(1)
```

```python
gp = GProfiler(
    return_dataframe=True, #return pandas dataframe or plain python structures    
)
```

```python
gp_result = gp.profile(organism='hsapiens',
                      query=top_genes['Name'].to_list())
```

```python
gp_result.to_csv("../output/enrichment_res.tsv", sep="\t")
```

```python
output_file = "../output/enrichment_res.xlsx"
sheet_column = "source"

with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
    for sheet_name, group in gp_result.groupby(sheet_column):
        sheet_name = re.sub(r'[\[\]\:\*\?\/\\]', '_', sheet_name)
        group.drop(columns=[sheet_column]).to_excel(
            writer,
            sheet_name=str(sheet_name)[:31],  # Excel limit: 31 chars
            index=False
        )
```
